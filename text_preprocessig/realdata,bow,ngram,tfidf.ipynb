```python
import nltk
```

```python
import pandas as pd
```

```python
# reading file
with open("../data/Complete_data .txt","r") as file:
    corpus = file.read()
```

```python
corpus = corpus.lower()
```

```python
documents = nltk.sent_tokenize(corpus)
print(documents)
```

```python
# apply regex for the only text
import re
for i in range(len(documents)):
    documents[i] = re.sub("[^a-zA-Z]"," ",documents[i])
```

```python
li = nltk.word_tokenize(documents[0])
li
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lem = WordNetLemmatizer()
final = []
for i in range(len(documents)):
    sentence = documents[i]
    words = nltk.word_tokenize(sentence)
    stop_words = [lem.lemmatize(word) for word in words if word not in stopwords.words('english')]
    stop_words = ' '.join(stop_words)
    final.append(stop_words)
    
```

```python
final
```

```python
len(final)
```

```python
# bagof words
from sklearn.feature_extraction.text import CountVectorizer

cv  = CountVectorizer(binary = False,ngram_range = (1,2))

x= cv.fit_transform(final)
```

```python
x[0].toarray()
```

```python
df = pd.DataFrame(x.toarray(),columns = cv.get_feature_names_out())
```

```python
df
```

```python
# Tf-idf 
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(ngram_range=(1,2))

data = tfidf.fit_transform(final)
```

```python
data.toarray()
```

```python
data = pd.DataFrame(data.toarray(),columns = tfidf.get_feature_names_out())
```
