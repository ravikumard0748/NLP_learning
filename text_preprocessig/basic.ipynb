## custom tokenization using Regex



```python
import re

def custom_tokenizer(text):
    # Regex pattern to match words, numbers, emojis, punctuation
    pattern = r'\w+|[^\w\s]'  # words (\w+), or punctuation ([^\w\s])
    tokens = re.findall(pattern, text)
    return tokens

# Test text
text = "I love NLP! 😍 It's amazing, isn't it? Let's build custom-tokenizers."

tokens = custom_tokenizer(text)
print(tokens)
```

## Stop words and stemming 

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
sentences = nltk.sent_tokenize(paragraph)
stemmer = PorterStemmer()
```

```python
for i in range(len(sentences)):
    word = sentences[i]
    words = nltk.word_tokenize(word)
    sentence  = [stemmer.stem(word) for word in words if word not in stopwords.words('english')]
    sentences[i] = ' '.join(sentence)
```
